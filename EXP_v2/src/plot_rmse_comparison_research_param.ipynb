```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

# notebook は EXP_v2/src/ にあるので parents[1] がプロジェクトルート
ROOT = Path().resolve().parents[1]
RESULTS_DIR = ROOT / "EXP_v2" / "results"
print(f"Project root: {ROOT}")
print(f"Results dir : {RESULTS_DIR}")
```

```python
dqn_normal_path       = RESULTS_DIR / "summary_2pl_1_100items_DQN_normal_epsilon_0.5_gamma_0.5_research_param.csv"
dqn_normal_fp64_path  = RESULTS_DIR / "summary_2pl_1_100items_DQN_normal_epsilon_0.3_gamma_0.0_research_param_fp64.csv"
dqn_uniform_path      = RESULTS_DIR / "summary_2pl_1_100items_DQN_uniform_epsilon_0.3_gamma_0.0_research_param.csv"
mfi_py_path           = RESULTS_DIR / "summary_2pl_1_100items_MFI_MLE_python.csv"

dqn_normal      = pd.read_csv(dqn_normal_path)
dqn_normal_fp64 = pd.read_csv(dqn_normal_fp64_path)
dqn_uniform     = pd.read_csv(dqn_uniform_path)
mfi_py          = pd.read_csv(mfi_py_path)

print(f"DQN (normal, fp32)  : {dqn_normal_path.name}")
display(dqn_normal)
print(f"DQN (normal, fp64)  : {dqn_normal_fp64_path.name}")
display(dqn_normal_fp64)
print(f"DQN (uniform, fp32) : {dqn_uniform_path.name}")
display(dqn_uniform)
print(f"MFI (Python)        : {mfi_py_path.name}")
display(mfi_py)

print(f"\nmean RMSE (steps 1-{len(dqn_normal)}):")
print(f"  DQN (normal, fp32)  : {dqn_normal['RMSE'].mean():.4f}")
print(f"  DQN (normal, fp64)  : {dqn_normal_fp64['RMSE'].mean():.4f}")
print(f"  DQN (uniform, fp32) : {dqn_uniform['RMSE'].mean():.4f}")
print(f"  MFI (Python)        : {mfi_py['RMSE'].mean():.4f}")
print(f"step {int(dqn_normal['step'].iloc[-1])} RMSE:")
print(f"  DQN (normal, fp32)  : {dqn_normal['RMSE'].iloc[-1]:.4f}")
print(f"  DQN (normal, fp64)  : {dqn_normal_fp64['RMSE'].iloc[-1]:.4f}")
print(f"  DQN (uniform, fp32) : {dqn_uniform['RMSE'].iloc[-1]:.4f}")
print(f"  MFI (Python)        : {mfi_py['RMSE'].iloc[-1]:.4f}")
```

```python
def plot_mfi_vs_dqns(dqn_curves, out_name, title):
    """dqn_curves: list of dicts with keys: df, label, color, marker."""
    fig, ax = plt.subplots(figsize=(8, 5))

    ax.plot(mfi_py["step"], mfi_py["RMSE"], label="MFI (Python, MLE)", color="steelblue", linewidth=1.8, marker="o")
    x_maxes = [mfi_py["step"].max()]
    for cv in dqn_curves:
        ax.plot(
            cv["df"]["step"], cv["df"]["RMSE"],
            label=cv["label"], color=cv["color"], linewidth=1.8, marker=cv["marker"],
        )
        x_maxes.append(cv["df"]["step"].max())

    ax.set_xlabel("Step", fontsize=13)
    ax.set_ylabel("RMSE", fontsize=13)
    ax.set_title(title, fontsize=13)
    ax.legend(fontsize=11)
    ax.grid(True, linestyle="--", alpha=0.5)
    ax.set_xlim(1, max(x_maxes))

    plt.tight_layout()

    out_path = RESULTS_DIR / out_name
    fig.savefig(out_path, dpi=150)
    plt.show()
    print(f"Saved to {out_path}")


plot_mfi_vs_dqns(
    dqn_curves=[
        {"df": dqn_normal,      "label": "DQN (normal, fp32, epsilon=0.5, gamma=0.5)", "color": "darkorange",    "marker": "s"},
        {"df": dqn_normal_fp64, "label": "DQN (normal, fp64, epsilon=0.3, gamma=0.0)", "color": "purple", "marker": "^"},
    ],
    out_name="rmse_comparison_2pl_1_100items_MFI_vs_DQN_normal.png",
    title="RMSE Comparison: MFI vs DQN (normal, fp32 / fp64) (EXP_v2, 2PL bank 1, 100 items)",
)

plot_mfi_vs_dqns(
    dqn_curves=[
        {"df": dqn_uniform, "label": "DQN (uniform, epsilon=0.3, gamma=0.0)", "color": "darkorange", "marker": "D"},
    ],
    out_name="rmse_comparison_2pl_1_100items_MFI_vs_DQN_uniform.png",
    title="RMSE Comparison: MFI vs DQN (uniform) (EXP_v2, 2PL bank 1, 100 items)",
)
```
